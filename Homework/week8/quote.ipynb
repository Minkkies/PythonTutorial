```python
import requests
```

```python
url = "https://quotes.toscrape.com/"
```

```python
response = requests.get(url) # ส่งคำขอไปยัง URL และเก็บผลลัพธ์ในตัวแปร response
```

```python
print(response.status_code) # แสดงเพื่อเช้คว่าได้ข้อมูลหรือไม่
```

```python
print(response.text) # แสดงข้อมูลที่ได้มา
```

```python
from bs4 import BeautifulSoup
```

```python
soup = BeautifulSoup(response.text, 'html.parser') # สร้างวัตถุ BeautifulSoup จาก HTML ที่ได้มา
```

```python
quotes = soup.find_all(class_='text') # ค้นหาทุก div ที่มี class เป็น text ซึ่งเป็นที่เก็บคำคม
```

```python
print(len(quotes)) # แสดงจำนวนคำคมที่พบ
print(quotes[0].text) # แสดงคำคมแรกที่พบ
```

```python
quotes_list =[]
for quote in quotes:
    print(quote.text) # แสดงคำคมแต่ละอันที่พบ
    quotes_list.append(quote.text) # เก็บคำคมใน list
```

```python
authors = soup.find_all(class_='author') # ค้นหาทุก div ที่มี class เป็น author ซึ่งเป็นที่เก็บชื่อผู้แต่ง
```

```python
authors_list =[]
for author in authors:
    print(author.text) # แสดงชื่อผู้แต่งแต่ละคนที่พบ
    authors_list.append(author.text) # เก็บชื่อผู้แต่งใน list
```

```python
authors_list
```

```python
# รวมกันเป็น Dictionary เดียว
data = {
'คำคม': quotes_list,
'ผู้เขียน': authors_list
}
```
